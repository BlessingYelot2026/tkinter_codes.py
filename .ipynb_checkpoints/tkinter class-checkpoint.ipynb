```python
import tkinter as tk
# 1. Create the main window
root = tk.Tk()
# 2. Give it a title and size
root.title("My First App")
#width X height in pixels
root.geometry("1000x600")  
# 3. Keep the window open and listening for events
root.mainloop()


#label
import tkinter as tk
root = tk.Tk()
root.title("Label Example")
root.geometry("400x200")
label = tk.Label(
    root,
    text="Hello, World!",
    font=("Arial", 16),
    fg="blue",         
    # text color (foreground)
    bg="lightyellow"   
    # background color
)
label.pack(pady=30)     
# pady adds vertical space
root.mainloop()


#Botton
import tkinter as tk
def say_hello():
    print("Button was clicked!")  
    # appears in the terminal
    label.config(text="Hello! You clicked me.")
root = tk.Tk()
root.title("Button Example")
root.geometry("350x200")
label = tk.Label(root, text="Click the button below", font=("Arial", 12))
label.pack(pady=20)
button = tk.Button(
    root,
    text="Click Me",
    command=say_hello,     
    # function to run when clicked
    font=("Arial", 11),
    bg="lightgreen",
    width=12
)
button.pack(pady=10)
root.mainloop()


#Entry(it create box for user to imput)
import tkinter as tk
def show_name():
    name = entry.get()                 # get the text
    result.config(text=f"Hello, {name}!")
root = tk.Tk()
root.title("Entry Example")
root.geometry("350x220")
tk.Label(root, text="What is your name?", font=("Arial", 12),bg="blue").pack(fill=tk.X)      
entry = tk.Entry(root, font=("Arial", 12), width=25)
entry.pack(pady=5)
entry.focus()                          # cursor starts here
tk.Button(root, text="Say Hello", command=show_name).pack(pady=10)
result = tk.Label(root, text="", font=("Arial", 14), fg="darkgreen")
result.pack(pady=10)


from tkinter import messagebox
# Information
messagebox.showinfo("Title", "This is useful information.")
# Warning
messagebox.showwarning("Warning", "Something needs attention.")
# Error
messagebox.showerror("Error", "Something went wrong.")
# Yes / No question
answer = messagebox.askyesno("Confirm", "Do you want to continue?")
if answer:
    print("User clicked Yes")
else:
    print("User clicked No")


    
#Frame
     import tkinter as tk
root = tk.Tk()
root.title("Frames")
root.geometry("400x250")
# Top frame for title
top = tk.Frame(root, bg="lightblue", height=50)
top.pack(fill=tk.X)
tk.Label(top, text="My App", bg="lightblue", font=("Arial", 16)).pack(pady=10)
# Middle frame for content
middle = tk.Frame(root)
middle.pack(pady=20)
tk.Button(middle, text="Button 1").pack(side=tk.LEFT, padx=10)
tk.Button(middle, text="Button 2").pack(side=tk.LEFT, padx=10)
root.mainloop()


#Exercise function
import tkinter as tk
from tkinter import messagebox

def create_fields():
    # Remove any previous input fields
    for widget in fields_frame.winfo_children():
        widget.destroy()
    entry_list.clear()

    try:
        n = int(entry_n.get())
        if n <= 0:
            messagebox.showerror("Error", "Please enter a positive number.")
            return
        if n > 5:
            messagebox.showerror("Error", "Maximum 5 numbers allowed.")
            return
    except ValueError:
        messagebox.showerror("Error", "Please enter a valid whole number.")
        return

    # Create the entry fields on the same window
    for i in range(n):
        row = tk.Frame(fields_frame)
        row.pack(pady=4)

        tk.Label(row, text=f"Value {i+1}:", width=10, anchor="e",
                 font=("Arial", 10)).pack(side=tk.LEFT)

        e = tk.Entry(row, width=12, font=("Arial", 11), justify="center")
        e.pack(side=tk.LEFT, padx=6)
        entry_list.append(e)

def calculate_average():
    if not entry_list:
        messagebox.showerror("Error", "Click 'Create Fields' first.")
        return

    values = []
    try:
        for e in entry_list:
            values.append(float(e.get()))
    except ValueError:
        messagebox.showerror("Error", "Please enter valid numbers in all fields.")
        return

    average = sum(values) / len(values)
    result_label.config(text=f"Average = {average:.2f}")

# ===================== Main Window =====================
root = tk.Tk()
root.title("Average Calculator")
root.geometry("400x520")
root.resizable(False, False)

# Title bar
top = tk.Frame(root, bg="lightblue", height=50)
top.pack(fill=tk.X)
tk.Label(top, text="Average Calculator", bg="lightblue",
         font=("Arial", 16, "bold")).pack(pady=12)

# How many numbers section
section1 = tk.Frame(root, pady=12)
section1.pack()

tk.Label(section1, text="How many numbers?", font=("Arial", 11)).pack()
entry_n = tk.Entry(section1, width=8, font=("Arial", 12), justify="center")
entry_n.pack(pady=4)

tk.Button(section1, text="Create Fields", command=create_fields,
          bg="#2196F3", fg="white", font=("Arial", 10),
          padx=12, pady=4).pack(pady=6)

# This frame will hold the dynamic Value 1, Value 2, ... entries
fields_frame = tk.Frame(root)
fields_frame.pack(pady=8)

entry_list = []   # stores the Entry widgets

# Calculate button
tk.Button(root, text="Calculate Average", command=calculate_average,
          bg="#4CAF50", fg="white", font=("Arial", 11, "bold"),
          padx=15, pady=6).pack(pady=12)

# Result label
result_label = tk.Label(root, text="Average will appear here",
                        bg="#e8f5e9", font=("Arial", 13),
                        width=28, pady=8)
result_label.pack(pady=8)

root.mainloop()
```

```python
#temperature converter
import tkinter as tk
from tkinter import messagebox
def convert():
    try:
        c = float(entry.get())
        f = c * 9/5 + 32
        result.config(text=f"{f:.1f} °F")
    except ValueError:
      label.config(text="Error,enter the right value")
root = tk.Tk()
root.title("Celsius fi Fahrenheit")
root.geometry("320x200")
tk.Label(root, text="Temperature in Celsius:").pack(pady=15)
entry = tk.Entry(root, width=15, font=("Arial", 14))
entry.pack()
entry.focus()
tk.Button(root, text="Convert", command=convert).pack(pady=15)
result = tk.Label(root, text="", font=("Arial", 18), fg="darkgreen")
result.pack()
label=tk.Label(root, text="").pack(pady=6)
root.mainloop()

#Counter
import tkinter as tk
count = 0
def increase():
    global count
    count += 1
    label.config(text=str(count))
def decrease():
    global count
    count -= 1
    label.config(text=str(count))
def reset():
    global count
    count = 0
    label.config(text="0")
def double():
    global count
    count *= 2
    label.config(text=str(count))
root = tk.Tk()
root.title("Counter + Double")
root.geometry("320x250")
label = tk.Label(root, text="0", font=("Arial", 48))
label.pack(pady=20)
btn_frame = tk.Frame(root)
btn_frame.pack()
tk.Button(btn_frame, text="+", width=5, command=increase).pack(side=tk.LEFT, padx=5)
tk.Button(btn_frame, text="-", width=5, command=decrease).pack(side=tk.LEFT, padx=5)
tk.Button(btn_frame, text="×2", width=5, command=double).pack(side=tk.LEFT, padx=5)
tk.Button(root, text="Reset", command=reset).pack(pady=15)
root.mainloop()
```

```python
import tkinter as tk
from tkinter import ttk
import csv
import os
import math
from datetime import datetime
# Name of the CSV file
CSV_FILE = "inventory_purchases.csv"
# Column headings for the CSV file
HEADERS = [
    "Date and Time",
    "Product Name",
    "Quantity",
    "Unit Price (NGN)",
    "Total Cost (NGN)"
]
# 1. Create the CSV file if it does not exist
def create_csv_file():
    if not os.path.exists(CSV_FILE) or os.path.getsize(CSV_FILE) == 0:
        with open(CSV_FILE, "w", newline="", encoding="utf-8-sig") as file:
            writer = csv.DictWriter(file, fieldnames=HEADERS)
            writer.writeheader()
# 2. Clear the input fields
def clear_fields():
    entry_name.delete(0, tk.END)
    entry_quantity.delete(0, tk.END)
    entry_price.delete(0, tk.END)
    entry_name.focus_set()
    status_label.config(
        text="Fields cleared. Saved records are safe.",
        fg="blue"
    )
# 3. Display saved records in the table
def refresh_records():
    # Remove old rows from the table
    for row in records_table.get_children():
        records_table.delete(row)
    total_spent = 0.0
    record_count = 0
    try:
        create_csv_file()
        with open(CSV_FILE, "r", newline="", encoding="utf-8-sig") as file:
            reader = csv.DictReader(file)
            for record in reader:
                date_time = record.get("Date and Time", "")
                name = record.get("Product Name", "")
                quantity = record.get("Quantity", "")
                unit_price = float(
                    record.get("Unit Price (NGN)", "0") or 0
                )
                total_cost = float(
                    record.get("Total Cost (NGN)", "0") or 0
                )
                # Add this record to the table
                records_table.insert(
                    "",
                    tk.END,
                    values=(
                        date_time,
                        name,
                        quantity,
                        f"{unit_price:,.2f}",
                        f"{total_cost:,.2f}"
                    )
                )
                # Add the cost to the total expenditure
                total_spent += total_cost
                record_count += 1
        total_label.config(
            text=f"Total recorded purchases: ₦{total_spent:,.2f} "
                 f"| Records: {record_count}"
        )
        status_label.config(
            text=f"Loaded {record_count} purchase record(s).",
            fg="blue"
        )
    except (OSError, ValueError, csv.Error) as error:
        status_label.config(
            text=f"Could not load records: {error}",
            fg="red"
        )
# 4. Validate and save a new purchase
def save_purchase():
    # Get the information entered by the user
    name = entry_name.get().strip()
    quantity_text = entry_quantity.get().strip()
    price_text = entry_price.get().strip().replace(",", "")
    # Check the product name
    if not name:
        status_label.config(
            text="Please enter a product name.",
            fg="red"
        )
        return
    # Convert quantity into an integer
    try:
        quantity = int(quantity_text)
    except ValueError:
        status_label.config(
            text="Quantity must be a whole number, such as 2 or 10.",
            fg="red"
        )
        return
    if quantity <= 0:
        status_label.config(
            text="Quantity must be greater than zero.",
            fg="red"
        )
        return
    # Convert price into a number
    try:
        unit_price = float(price_text)
    except ValueError:
        status_label.config(
            text="Enter a valid price, such as 2500 or 2500.50.",
            fg="red"
        )
        return
    if not math.isfinite(unit_price) or unit_price < 0:
        status_label.config(
            text="Price must be a valid, non-negative number.",
            fg="red"
        )
        return
    # Calculate the total purchase cost
    total_cost = quantity * unit_price
    # Put the purchase information into a dictionary
    record = {
        "Date and Time": datetime.now().strftime(
            "%Y-%m-%d %H:%M:%S"
        ),
        "Product Name": name,
        "Quantity": quantity,
        "Unit Price (NGN)": f"{unit_price:.2f}",
        "Total Cost (NGN)": f"{total_cost:.2f}"
    }
    # Save the record in the CSV file
    try:
        create_csv_file()
        with open(CSV_FILE, "a", newline="", encoding="utf-8") as file:
            writer = csv.DictWriter(file, fieldnames=HEADERS)
            writer.writerow(record)
        # Update the table
        refresh_records()
        # Clear the input boxes
        entry_name.delete(0, tk.END)
        entry_quantity.delete(0, tk.END)
        entry_price.delete(0, tk.END)
        entry_name.focus_set()
        # Show a success message inside the window
        status_label.config(
            text=f"Saved {name}: {quantity} unit(s), "
                 f"total ₦{total_cost:,.2f}.",
            fg="green"
        )
    except (OSError, csv.Error) as error:
        status_label.config(
            text=f"Could not save purchase: {error}",
            fg="red"
        )
# 5. Create the main window
root = tk.Tk()
root.title("Beginner Inventory - Goods Purchased")
root.geometry("900x600")
root.minsize(760, 500)
# Application title
title_label = tk.Label(
    root,
    text="Mini Inventory - Goods Purchased",
    font=("Arial", 16, "bold")
)
title_label.pack(pady=12)
# 6. Create the input form
form_frame = tk.LabelFrame(
    root,
    text="Enter a new purchase",
    padx=10,
    pady=10
)
form_frame.pack(fill="x", padx=12, pady=5)
# Product name
tk.Label(form_frame, text="Product name:").grid(
    row=0, column=0, sticky="w", padx=5, pady=6
)
entry_name = tk.Entry(form_frame, width=25)
entry_name.grid(
    row=0, column=1, sticky="ew", padx=5, pady=6
)
# Quantity
tk.Label(form_frame, text="Quantity bought:").grid(
    row=0, column=2, sticky="w", padx=5, pady=6
)
entry_quantity = tk.Entry(form_frame, width=15)
entry_quantity.grid(
    row=0, column=3, sticky="ew", padx=5, pady=6
)
# Unit price
tk.Label(form_frame, text="Unit price (NGN):").grid(
    row=1, column=0, sticky="w", padx=5, pady=6
)
entry_price = tk.Entry(form_frame, width=25)
entry_price.grid(
    row=1, column=1, sticky="ew", padx=5, pady=6
)
# Allow input columns to expand
form_frame.columnconfigure(1, weight=1)
form_frame.columnconfigure(3, weight=1)
# 7. Create the buttons
button_frame = tk.Frame(root)
button_frame.pack(fill="x", padx=12, pady=8)
tk.Button(
    button_frame,
    text="Save Purchase",
    command=save_purchase,
    bg="#2E8B57",
    fg="white",
    padx=10,
    pady=5
).pack(side="left", padx=4)
tk.Button(
    button_frame,
    text="Clear Fields",
    command=clear_fields,
    padx=10,
    pady=5
).pack(side="left", padx=4)
tk.Button(
    button_frame,
    text="Refresh Records",
    command=refresh_records,
    padx=10,
    pady=5
).pack(side="left", padx=4)
# 8. Create the records table
table_frame = tk.LabelFrame(
    root,
    text="Saved purchase records",
    padx=6,
    pady=6
)
table_frame.pack(
    fill="both", expand=True, padx=12, pady=5
)
columns = (
    "date",
    "product",
    "quantity",
    "unit_price",
    "total"
)
records_table = ttk.Treeview(
    table_frame,
    columns=columns,
    show="headings"
)
# Table headings
records_table.heading("date", text="Date and Time")
records_table.heading("product", text="Product Name")
records_table.heading("quantity", text="Quantity")
records_table.heading("unit_price", text="Unit Price (₦)")
records_table.heading("total", text="Total Cost (₦)")
# Table column sizes and alignment
records_table.column("date", width=145, anchor="w")
records_table.column("product", width=200, anchor="w")
records_table.column("quantity", width=80, anchor="center")
records_table.column("unit_price", width=130, anchor="e")
records_table.column("total", width=130, anchor="e")
# Scrollbar
scrollbar = ttk.Scrollbar(
    table_frame,
    orient="vertical",
    command=records_table.yview
)
records_table.configure(
    yscrollcommand=scrollbar.set
)
records_table.pack(
    side="left", fill="both", expand=True
)
scrollbar.pack(side="right", fill="y")
# 9. Display total expenditure
total_label = tk.Label(
    root,
    text="Total recorded purchases: ₦0.00",
    font=("Arial", 10, "bold")
)
total_label.pack(anchor="e", padx=14, pady=4)
# 10. Display status and error messages
status_label = tk.Label(
    root,
    text="Ready. Enter a purchase to begin.",
    fg="blue",
    wraplength=850
)
status_label.pack(fill="x", padx=12, pady=8)
# Create the CSV file and display any existing records
create_csv_file()
refresh_records()
# Keep the window running
root.mainloop()
   
```
